## Cell 2: Import Libraries & Initialize Pipeline
- Imports standard audio processing libraries (`soundfile`, `numpy`, `IPython.display.Audio`).
- Initializes `KPipeline` with language code (`'a'` for American English, `'b'` for British English).

```python
from kokoro import KPipeline
import soundfile as sf
import numpy as np
from IPython.display import display, Audio

# Initialize pipeline (lang_code='a' for American English)
pipeline = KPipeline(lang_code='a')
print("Kokoro Pipeline ready.")
```

## Cell 3: Speech Generation & Audio Export
- Selects the target voice (e.g., `'af_bella'`).
- Passes input text to the generator model.
- Concatenates generated audio segments into a continuous audio file (`final_output.wav`).
- Plays back the generated audio directly in the notebook.

```python
chosen_voice = 'af_bella'

text = """
Hello! Welcome to the Kokoro Text-to-Speech demonstration notebook. This systematic setup generates high quality audio.
"""

generator = pipeline(text, voice=chosen_voice, speed=1.0, split_pattern=r'\n+')
all_audio = []
for i, (gs, ps, audio) in enumerate(generator):
    all_audio.append(audio)
final_audio = np.concatenate(all_audio)
sf.write("final_output.wav", final_audio, 24000)
print("Success! Saved ONE single file: final_output.wav")
display(Audio(data=final_audio, rate=24000, autoplay=True))
```
